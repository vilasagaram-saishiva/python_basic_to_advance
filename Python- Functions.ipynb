```python
def create_table_name(database,table,environment = 'dev'):
    return(f'{environment}.{table}')

print(create_table_name("my_database","customer"))
print(create_table_name("my_database","client","prod"))
```

```python
def add_numbers(*args):
    total = 0
    for number in args:
        total = total + number

    return total
add_numbers(10,40)
```

```python
def greet_people(greeting, *names):
    for name in names:
        print(greeting, name)
```

```python
greet_people("Hello", "Rahul", "Amit", "John")
```
